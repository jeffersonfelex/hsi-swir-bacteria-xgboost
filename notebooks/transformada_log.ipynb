```python
import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as plt
```

```python
input_csv = "/mnt/c/Users/jefferson.faria/Documents/GitHub/ic-2025/ETL/data_processed/dados_corregidos_msc.csv"
output_csv = "/mnt/c/Users/jefferson.faria/Documents/GitHub/ic-2025/ETL/data_processed/dados_transformada_log.csv"
```

```python
df = pd.read_csv(input_csv)
data = df.values
```

```python
# Evita log de zero ou valores negativos
data[data <= 0] = 1e-6
data_log = -np.log(data)
```

```python
# Espectos original
plt.figure(figsize=(10, 4))
for i in range(min(10, data.shape[0])):  # Plota até 10 amostras
    plt.plot(data[i], alpha=0.5)
plt.title("Espectros Originais")
plt.xlabel("Bandas")
plt.ylabel("Reflectância")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Espectros após log
plt.figure(figsize=(10, 4))
for i in range(min(10, data_log.shape[0])):
    plt.plot(data_log[i], alpha=0.5)
plt.title("Espectros após Transformada Logarítmica (log(1/R))")
plt.xlabel("Bandas")
plt.ylabel("-log(Reflectância)")
plt.grid(True)
plt.tight_layout()
plt.show()
```
