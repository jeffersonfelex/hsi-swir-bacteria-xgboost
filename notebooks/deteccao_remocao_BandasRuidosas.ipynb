```python
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
input_csv = "/mnt/c/Users/jefferson.faria/Documents/GitHub/ic-2025/ETL/data_processed/dados_derivada_primeira_ordem.csv"
output_csv = "/mnt/c/Users/jefferson.faria/Documents/GitHub/ic-2025/ETL/data_processed/dados_sem_bandas_ruidosas.csv"
```

```python
df = pd.read_csv(input_csv)

limiar_var=1e-6
variancias = df.var()

bandas_boas = variancias[variancias > limiar_var].index
df_filtrado = df[bandas_boas]
```

```python
plt.figure(figsize=(12, 5))
plt.plot(df.values[0], label="Original", alpha=0.6)
plt.plot(df_filtrado.values[0], label="Sem bandas ruidosas", alpha=0.8)
plt.title("Comparação: Espectro com e sem bandas ruidosas")
plt.xlabel("Bandas")
plt.ylabel("Valor")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
os.makedirs(os.path.dirname(output_path), exist_ok=True)
df_filtrado.to_csv(output_path, index=False)
```
