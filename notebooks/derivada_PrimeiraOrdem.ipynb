```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter
import os
```

```python
input_csv = "/mnt/c/Users/jefferson.faria/Documents/GitHub/ic-2025/ETL/data_processed/dados_transformada_log.csv"
output_csv = "/mnt/c/Users/jefferson.faria/Documents/GitHub/ic-2025/ETL/data_processed/dados_derivada_primeira_ordem.csv"
```

```python
df = pd.read_csv(input_csv)
X = df.values

derivada = savgol_filter(X, window_length=11, polyorder=2, deriv=1)
```

```python
plt.figure(figsize=(12, 5))
plt.plot(X[0], label="Original", linewidth=2)
plt.plot(derivada[0], label="Derivada 1ª Ordem", linewidth=2)
plt.title("Comparação: Sinal Original vs Derivada de 1ª Ordem")
plt.xlabel("Bandas Espectrais")
plt.ylabel("Valor")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
df_derivada = pd.DataFrame(derivada, columns=df.columns)
os.makedirs(os.path.dirname(output_csv), exist_ok=True)
df_derivada.to_csv(output_csv, index=False)
```
