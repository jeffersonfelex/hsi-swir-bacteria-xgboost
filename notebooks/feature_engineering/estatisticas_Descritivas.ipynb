```python
from scipy.stats import kurtosis, skew
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import os
```

```python
input_path = "/mnt/c/Users/jefferson.faria/Documents/GitHub/ic-2025/dataSet/data-preProcessed/pre_processing/espectros_bandas_filtradas.csv"
output_dir = "/mnt/c/Users/jefferson.faria/Documents/GitHub/ic-2025/dataSet/data-preProcessed/pre_processing"
output_path = os.path.join(output_dir, 'espectros_estatisticas_Descritivas.csv')
```

```python
df = pd.read_csv(input_path)
data = df.values
```

```python
stats = pd.DataFrame(index=df.index)
stats['media'] = np.mean(data, axis=1)
stats['desvio_padrao'] = np.std(data, axis=1, ddof=1)  # ddof=1 para amostra
stats['curtose'] = [kurtosis(row, fisher=True) for row in data]  # Fisher=True -> normal=0
stats['skewness'] = [skew(row) for row in data]
```

```python
plt.figure(figsize=(14,6))
plt.plot(stats['media'], label='Média', marker='o')
plt.plot(stats['desvio_padrao'], label='Desvio Padrão', marker='x')
plt.plot(stats['curtose'], label='Curtose', marker='s')
plt.plot(stats['skewness'], label='Skewness', marker='^')
plt.title("Estatísticas Descritivas por Amostra/Espectro")
plt.xlabel("Amostra/Espectro")
plt.ylabel("Valor")
plt.legend()
plt.grid(True)
plt.show()
```
